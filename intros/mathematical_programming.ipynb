## Problem 6. Simple Linear Regression

Given a set of points $(x_1, y_1), (x_2, y_2), \ldots, (x_n, y_n)$, $n>1$, with distinct values for $x$ find the _best fit line*_ that minimizes the mean squared error:
$$
\text{MSE}(k, b) = \frac{1}{n}\sum_{i=1}^{n} (kx_i + b - y_i)^2.
$$
In other words, find
$$
k, b = \argmin_{k, b} \text{MSE}(k, b).
$$
*the _best fit line_ is given by the equation $y = kx + b$ where $k$ is the slope and $b$ is the y-intercept.

````python
import numpy as np


def best_line_np(points: list[tuple[float, float]]) -> tuple[float, float]:
    """The best line using numpy.polyfit.
    Use this function to test your implementation."""
    xs, ys = zip(*points)
    coefs = np.polyfit(xs, ys, 1)
    return coefs[0], coefs[1]


def generate_points(k: float, b: float, n: int) -> list[tuple[float, float]]:
    """Generate n points close to the line y = kx + b."""
    xs = np.random.random(n) * 10 - 5
    xs.sort()
    ys = k * xs + b + np.random.randn(n)
    return list(zip(xs, ys))


def best_line(points: list[tuple[float, float]]) -> tuple[float, float]:
    ...
````
