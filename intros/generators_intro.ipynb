# generator objects

For simplicity, let's define a generator object as an object that can be iterated over, and that generates its elements on the fly, rather than storing them in memory.

For example, the following two functions are equivalent, but the first one uses a list comprehension to generate all the elements at once, while the second one uses a generator object (using generator comprehension syntax) to generate the elements one by one when they are requested by the `sum` function.

```python
def sum_squares_list(n: int) -> int:
    values = [i**2 for i in range(1, n + 1)]
    return sum(values)


def sum_squares_gen(n: int) -> int:
    values = (i**2 for i in range(1, n + 1))
    return sum(values)


print(sum_squares_list(10))
print(sum_squares_gen(10))
```

The first function has to allocate memory for all the elements at once, while the second function only needs to allocate memory for one element at a time. 

This can be a huge advantage when dealing with large datasets especially when one needs to iterate over the elements only once.

This is the main motivation for using generator objects.

Let's look at more syntax and examples of generator objects.

```python
from typing import Iterator, Generator


def squares(n: int) -> list[int]:
    """Return a list of the squares of the numbers from 1 to n inclusive."""
    return [i**2 for i in range(1, n + 1)]


def squares_gen_v1(n: int) -> Iterator[int]:
    """Return a generator of the squares of the numbers from 1 to n inclusive.
    
    This is a generator function that uses a generator comprehension.
    Returns a generator object directly.
    """
    return (i**2 for i in range(1, n + 1))


def squares_gen_v2(n: int) -> Generator[int, None, None]:
    """Generate the squares of the numbers from 1 to n inclusive.

    This is a generator function that uses the `yield` statement.
    This is more flexible than a generator comprehension in a sense that:
    - it can generate values in a loop
    - it can generate values indefinitely
    - it can recieve and return values (advanced)
    """
    for i in range(1, n + 1):
        yield i**2
```

All of these functions allow us to iterate over the given number of squares:

```python
N = 1_000_000
sq_lst = squares(N)
sq_gen_v1 = squares_gen_v1(N)
sq_gen_v2 = squares_gen_v2(N)

print(
    "All elements are the same when iterated over:",
    all(x == y == z for x, y, z in zip(sq_lst, sq_gen_v1, sq_gen_v2)),
)
```

However, when we look at the memory usage, we can see that the generator objects are much more memory efficient than the lists.

```python
import sys
print(sys.getsizeof(sq_lst))
print(sys.getsizeof(sq_gen_v1))
print(sys.getsizeof(sq_gen_v2))
```

Actually, when we iterate over the generator objects the `next` function is called, which returns the next element of the generator object.

You may also notice that many built-in functions in Python, such as `range`, `map`, `filter`, and `zip`, return generator objects. Not only that, the file streams in Python are also generator objects:

```python
import pathlib


FILES_DIR = pathlib.Path('pt3_files')


with open(FILES_DIR / "grades.jsonl") as f:
    print(f)
    print(next(f), end='')
    print(next(f), end='')
    print(next(f), end='')
    print(next(f), end='')

# or, more conventionally
with open(FILES_DIR / "csv_files" / "example.csv") as f:
    for line in f:
        print(line, end='')
```
