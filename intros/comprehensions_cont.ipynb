## `dict` comprehension

Same idea, different object.

Instead of
```python
d = {}
for k, v in some_iterable:
    d[k] = v
```

you can write
```python
d = {k: v for k, v in some_iterable}
```

````python
# example 1: map numbers to their squares

numbers = [1, 2, 3, 4, 5]

d1_1 = {}
for n in numbers:
    d1_1[n] = n**2
print(d1_1)

d1_2 = {n: n**2 for n in numbers}
print(d1_2)
````

````python
# example 2: map students to their grades
students = ['Alice', 'Bob', 'Charlie', 'David']
grades = [3.2, 3.5, 4.7, 4.0]

d2_1 = {}
for s, g in zip(students, grades):
    d2_1[s] = g
print(d2_1)

d2_2 = {s: g for s, g in zip(students, grades)}
print(d2_2)
````

One can also add conditions to the comprehension

Instead of 
```python
d = {}
for k, v in some_iterable:
    if some_condition(k, v):
        d[k] = v
```

you can write
```python
d = {k: v for k, v in some_iterable if some_condition(k, v)}
```

````python
# example 3: map students to their grades, but only for students with grades >= 4.0
d3_1 = {}
for s, g in zip(students, grades):
    if g >= 4.0:
        d3_1[s] = g
print(d3_1)

d3_2 = {s: g for s, g in zip(students, grades) if g >= 4.0}
print(d3_2)
````

## `set` comprehension

Same idea, different object.

````python
# example: unique products
number_pairs = [(3, 2), (4, 4), (2, 8), (2, 3), (7, 5), (6, 1), (0, 1), (4, 0)]

s_1 = set()
for a, b in number_pairs:
    s_1.add(a * b)
print(s_1)

s_2 = {a * b for a, b in number_pairs}
print(s_2)

assert s_1 == s_2
````
