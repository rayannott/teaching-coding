### All hashtags

You need to find all hashtags in a given text. A hashtag is a word that starts with a `#` symbol. The word can contain letters, digits, and underscores. The word ends when a non-letter, non-digit, and non-underscore character is encountered.

```python
# text containing hashtags
SAMPLE_TEXT = '''
#Python is an #awes0me language with a lot of features#.
This is a #sample_text which contains ##haSHtags.
Some more dummy text: #ab$cd_ #wall_e #wall-e
'''
# find_hashtags(SAMPLE_TEXT) should return:
# ['#Python', '#awes0me', '#sample_text', '#haSHtags', '#ab', '#wall_e', '#wall']
```

```python
def find_hashtags(text) -> list[str]:
    return re.findall(r'#\w+', text)
```

```python
find_hashtags(SAMPLE_TEXT)
```
